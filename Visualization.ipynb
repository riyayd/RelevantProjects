```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = [20,22,24,25,26] # marks of A in 5 sub
B = [17,20,19,24,23] # marks of B in 5 sub
C = [1,2,3,4,5]
```

```python
plt.scatter(C,A)
```

```python
plt.scatter(C,A, color = 'blue', label = 'Student A', marker = 'x')
plt.scatter(C,B, color = 'orange', label = 'Student B', marker = 'D')

plt.title('Students Performance Comparison')
plt.xlabel('Subject')
plt.ylabel('Marks')

plt.legend() # to display values written in label parameter legened is used
plt.grid() # to have backgorund lines
plt.show() # will eliminate this unwanted information
```

```python
plt.plot(C,A, label = 'Line plot', linestyle = '--', marker = 'o') # line plot

plt.title('Students Performance Comparison')
plt.xlabel('Subject')
plt.ylabel('Marks')

plt.legend() # to display values written in label parameter legened is used
plt.grid() # to have backgorund lines
plt.show() # will eliminate this unwanted information
```

```python
fig , axs = plt.subplots(1,2) # 1 row and 2 columns

axs[0].scatter(C,A) #  0 index (1st graph)
axs[0].set_ylabel('marks')

axs[1].scatter(C,B) # 1 index ( 2nd graph)
axs[1].set_xlabel('Subjects')
plt.show()
```

```python
df1 = pd.read_csv('auto-mpg.csv')
df1.head(3)
```

#### Bar Plot

```python
# Categorial data ( categories )
```

```python
counts = df1['origin'].value_counts() # will identify unique values in column and count their no.of appearances
counts
```

```python
x = list(counts.index)
x
```

```python
y = list(counts.values)
y
```

```python
plt.bar(x,y, color = ['red','black','green'])

plt.xlabel('Unique values')
plt.ylabel('Frequency')

plt.show()
```

#### Pie Plot

```python
# % --> percentages
# .2f --> represent numbers in float with 2 digits
```

```python
plt.pie(y , labels= x, autopct= '%.2f')
plt.show()
```
